```python
from typing import Optional
from pydantic import BaseModel, ValidationError

# Define the InferenceRequest class
class InferenceRequest(BaseModel):
    PassengerId: int
    Pclass: int
    Name: str
    Sex: str
    Age: Optional[float]
    SibSp: int
    Parch: int
    Ticket: str
    Fare: float
    Cabin: Optional[str]
    Embarked: str
    Title: str
    FamilySize: int
    IsAlone: int
    Deck: Optional[str]
    TicketGroupSize: int
    FarePerPerson: float
    AgeBin: Optional[int]
    FareBin: Optional[int]
    Pclass_AgeBin: Optional[float]
    Sex_Pclass: str
    CabinMissing: int
    AgeMissing: int

# Sample data
data = {
    "dataframe_split": {
        "columns": [
            "PassengerId", "Pclass", "Name", "Sex", "Age", "SibSp", "Parch", 
            "Ticket", "Fare", "Cabin", "Embarked", "Title", "FamilySize", "IsAlone", 
            "Deck", "TicketGroupSize", "FarePerPerson", "AgeBin", "FareBin", "Pclass_AgeBin", 
            "Sex_Pclass", "CabinMissing", "AgeMissing"
        ],
        "data": [
            [709, 3, "Moubarek, Master. Halim Gonios (\"William George\")", "male", 1, 1, 1, "2661", 
             15.2458, "c22", "C", "Master", 3, 0, "c", 2, 5.081933333333333, 1, 2, 1, "male_3", 1, 1],
            [709, 3, "Moubarek, Master. Halim Gonios (\"William George\")", "male", 1, 1, 1, "2661", 
             15.2458, "c22", "C", "Master", 3, 0, "c", 2, 5.081933333333333, 1, 2, 1, "male_3", 1, 1],
            [709, 3, "Moubarek, Master. Halim Gonios (\"William George\")", "male", 1, 1, 1, "2661", 
             15.2458, "c22", "C", "Master", 3, 0, "c", 2, 5.081933333333333, 1, 2, 1, "male_3", 1, 1]
        ]
    }
}

# Extract columns and data from the input dictionary
columns = data["dataframe_split"]["columns"]
rows = data["dataframe_split"]["data"]

# Convert each row into an InferenceRequest instance
inference_requests = []

for row in rows:
    # Map the row values to a dictionary using the column names
    row_dict = dict(zip(columns, row))
    
    try:
        # Create an InferenceRequest instance and append it to the list
        inference_request = InferenceRequest(**row_dict)
        inference_requests.append(inference_request)
    except ValidationError as e:
        print(f"Validation error for row {row}: {e}")

# Output the InferenceRequest instances
for request in inference_requests:
    print(request)
    
# convert the list of InferenceRequest instances to a dataframe
import pandas as pd
df = pd.DataFrame([request.dict() for request in inference_requests])
```
