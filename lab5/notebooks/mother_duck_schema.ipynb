```python
from dotenv import load_dotenv
import duckdb
import os

load_dotenv()
db_token = os.getenv("MOTHERDUCK_TOKEN")
conn = duckdb.connect(f"md:?motherduck_token={db_token}")
```

```python
# check if the connection is established
if conn:
    print("Connection to MotherDuck established successfully.")
```

```python
conn.execute("CREATE DATABASE IF NOT EXISTS weather_data")
conn.execute("USE weather_data")

# create a new table
conn.execute(
    """
    CREATE TABLE IF NOT EXISTS hourly_weather_data (
        timestamp TIMESTAMP,
        temperature FLOAT,
        relative_humidity FLOAT,
        rain FLOAT,
        precipitation FLOAT,
        cloud_cover FLOAT,
        wind_speed FLOAT
    )
"""
)
```

```python
conn.execute("USE weather_data")

conn.execute(
    """
    CREATE TABLE IF NOT EXISTS daily_weather_data (
        day_date DATE,
        temperature FLOAT,
        relative_humidity FLOAT,
        rain FLOAT,
        precipitation FLOAT,
        cloud_cover FLOAT,
        wind_speed FLOAT
    )
"""
)
```
